## "Distribution of Types of Disability Among Individuals"

```python
# Count
import matplotlib.pyplot as plt
import seaborn as sns
disability_counts = Individual01['12. Type of disability'].value_counts()
print(disability_counts)

# Plot
plt.figure(figsize=(10,5))
sns.barplot(x=disability_counts.index, y=disability_counts.values, palette='Set2')
plt.title('Type of Disability Count')
plt.xlabel('Type of Disability')
plt.ylabel('Number of Individuals')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
