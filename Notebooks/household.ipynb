Association Rule Mining for Housing Characteristics: Discovering Key Relationships in Household Data

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Select relevant columns for ARM
cols_for_arm = ['Q.1  Type of the house ownership', 'Q.2.1 Type of foundation of the house',
                'Q.2.2 Type of outer wall of the house', 'Q.2.3 Type of roof of the house']

# Drop rows with missing values in these columns
arm_data = Household_cleaned[cols_for_arm].dropna()

# Convert each row into a list of strings like ['Ownership=Own', 'Foundation=Mud bonded bricks/stone', ...]
transactions = arm_data.apply(lambda x: [f"{col}={val}" for col, val in x.items()], axis=1).tolist()

# One-hot encode the transactions
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_onehot = pd.DataFrame(te_ary, columns=te.columns_)

# Generate frequent itemsets with minimum support (adjust support threshold as needed)
frequent_itemsets = apriori(df_onehot, min_support=0.01, use_colnames=True)

# Generate association rules with min confidence threshold
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)

# Filter rules where antecedent or consequent involve roof type or ownership
filtered_rules = rules[
    (rules['antecedents'].apply(lambda x: any('Q.2.3 Type of roof' in i for i in x)) |
     rules['consequents'].apply(lambda x: any('Q.2.3 Type of roof' in i for i in x)) |
     rules['antecedents'].apply(lambda x: any('Q.1  Type of the house ownership' in i for i in x)) |
     rules['consequents'].apply(lambda x: any('Q.1  Type of the house ownership' in i for i in x)))
].sort_values(by='confidence', ascending=False)

filtered_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10)
```

```python
print(f"Total association rules generated: {len(rules)}")
```

```python
# Select the first rule
rule = rules.iloc[0]

# Print the rule details
print(f"Rule: {set(rule['antecedents'])} -> {set(rule['consequents'])}")
print(f"Support: {rule['support']:.4f}")
print(f"Confidence: {rule['confidence']:.4f}")
print(f"Lift: {rule['lift']:.4f}")
```

```python
# Sort rules by confidence in descending order
rules_sorted = rules.sort_values(by='confidence', ascending=False).head(5)

# Print top 5 rules
for i, rule in rules_sorted.iterrows():
    antecedents = ', '.join(list(rule['antecedents']))
    consequents = ', '.join(list(rule['consequents']))
    print(f"Rule {i+1}: If [{antecedents}] then [{consequents}]")
    print(f" - Support: {rule['support']:.4f}, Confidence: {rule['confidence']:.4f}, Lift: {rule['lift']:.4f}\n")
```

# Classification of house owner according to features

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder

# cleaned household dataframe is Household_cleaned
# and has columns renamed and mapped.

# Select features and target
features = [
    'Q.2.1 Type of foundation of the house',
    'Q.2.2 Type of outer wall of the house',
    'Q.2.3 Type of roof of the house',
    'Q.3 Main source of drinking water',
    'Q.4 Usual type of fuel used for cooking',
    'Q.5 Usual source of lighting',
    'Q.6 Type of toilet used',
    'Q.7.Household amenities - 1',  # You can add more amenities if available and cleaned
]

target = 'Q.1  Type of the house ownership'

# Filter to rows where none of these columns are null
df_model = Household_cleaned[features + [target]].dropna()

# Encode categorical variables with LabelEncoder
le_dict = {}
for col in features + [target]:
    le = LabelEncoder()
    df_model[col] = le.fit_transform(df_model[col])
    le_dict[col] = le  # save encoder for inverse transform later if needed

# Split data into train and test sets (70% train, 30% test)
X = df_model[features]
y = df_model[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize Decision Tree Classifier
clf = DecisionTreeClassifier(random_state=42)

# Train model
clf.fit(X_train, y_train)

# Predict on test data
y_pred = clf.predict(X_test)

# Evaluate performance
print("Classification Report:\n", classification_report(y_test, y_pred, target_names=le_dict[target].classes_))
```

```python
# Compute confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Get class names from label encoder
class_names = le_dict[target].classes_

# Plot confusion matrix heatmap
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix - Decision Tree Classifier')
plt.show()
```
